```python
df = df.drop(columns = "PassengerId", errors="ignore")
```

```python
cat_columns = ["Sex", "Pclass", "Embarked"]
df = pd.get_dummies(df, columns=cat_columns, drop_first=True)
```

```python
df
```

```python
result_col = ["Sex_male", "Pclass_2", "Pclass_3", "Embarked_Q", "Embarked_S"]
df[result_col] = df[result_col].astype(int)
```

```python
df
```

```python
from scipy.stats import chi2_contingency

categorical_cols = ['Sex_male', 'Pclass_2', 'Pclass_3', 'Embarked_Q', 'Embarked_S']

print("--- CHI-SQUARE TEST RESULTS ---")
cols_to_drop = []

for col in categorical_cols:
    contingency_table = pd.crosstab(df[col], df['Survived'])
    
    chi2, p, dof, expected = chi2_contingency(contingency_table)
    
    print(f"\nColumn: {col}")
    print(f"p-value: {p:.5e}")  
    
    if p >= 0.05:
        print("-> Decision: Faltu Column (Survived se koi gehra talluq nahi hai) - Drop it!")
        cols_to_drop.append(col)
    else:
        print("-> Decision: Important Column (Survived se gehra talluq hai) - Keep it!")

if cols_to_drop:
    df.drop(columns=cols_to_drop, inplace=True)
    print(f"\n✅ Dropped Columns: {cols_to_drop}")
else:
    print("\n✅ Koi bhi categorical column faltu nahi nikla! Saare columns model ke liye zaroori hain.")
```
